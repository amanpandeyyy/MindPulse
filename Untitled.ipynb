# Data Cleaning

```python
df = df.drop_duplicates()
# cov negative and unrealistic value to realistic value
df["Physical_Activity_Hours"]= df["Physical_Activity_Hours"].clip(lower = 0)
```

```python
df.describe()
```

```python
num_cols = df.select_dtypes(include = 'number')
num_cols.skew()
# near to zero => centralized (0.01 , 0.002)
# negative => left skewed(-1.56)
# positive => right skewed(1.256)
```

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_country(Country):
    if Country in top_countries:
        return Country
    else:
        return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_country)
```

```python
df['Grouped_country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
skewed_col = ['Study_Hours']
other_numeric_cols = ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks', 'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
normal_col = ['Gender', 'Academic_Level', 'Most_Used_Platform', 'Purpose_Of_Use', 'Grouped_country']

feature_col = skewed_col + other_numeric_cols + ordinal_col + normal_col
x = df[feature_col]
y = df['Mental_Health_Score']
X_train, X_test, y_train, y_test = train_test_split(
x, y, test_size=0.30, random_state=42
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OrdinalEncoder, OneHotEncoder
import numpy as np
from sklearn.compose import ColumnTransformer
# Skewed features pipeline
skew_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('scale', StandardScaler())
])

# Numeric features pipeline
plain_numeric_pipeline = Pipeline(steps=[
    ('scale', StandardScaler())
])

# Ordinal features pipeline
ordinal_pipeline = Pipeline(steps=[
    ('ordinal', OrdinalEncoder(categories=[['Low', 'Medium', 'High', 'Very High']]))
])

# Nominal features pipeline
nominal_pipeline = Pipeline(steps=[
    ('encode', OneHotEncoder(handle_unknown= 'ignore'))
])
#konsi pipeline mai konsa feture jana chaiye
preprocessor = ColumnTransformer(transformers =[
    ("Skewed_pipeline" , skew_pipeline, skewed_col),
    ("Plain_Numeric", plain_numeric_pipeline , other_numeric_cols),
    ("ordinal", ordinal_pipeline , ordinal_col),
    ('nominal', nominal_pipeline , normal_col),
])
```

```python
# Linear regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_absolute_error , mean_squared_error
lr_pipeline = Pipeline(steps = [
    ('preprocessor' , preprocessor),
    ('regressor', LinearRegression())
])
lr_pipeline.fit(X_train , y_train)
lr_preds = lr_pipeline.predict(X_test)
lr_preds_train = lr_pipeline.predict(X_train)

lr_r2_score_testing = r2_score(y_test , lr_preds)
lr_r2_score_traing = r2_score(y_train ,lr_preds_train)
lr_mae = mean_absolute_error(y_test , lr_preds)

print(f"Accuracy of training{lr_r2_score_traing }")
print(f"Accuracy of testing{lr_r2_score_testing}")
print(f"MAE{lr_mae}")
```

```python
# random_forest
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = rf_pipeline = Pipeline(steps = [
    ('preprocessor' , preprocessor),
    ('random forest', RandomForestRegressor(random_state = 42))
     ])
rf_pipeline.fit(X_train , y_train)
rf_preds = rf_pipeline.predict(X_test)
rf_preds_train = rf_pipeline.predict(X_train)

rf_r2_score_testing = r2_score(y_test , rf_preds)
rf_r2_score_traing = r2_score(y_train ,rf_preds_train)
rf_mae = mean_absolute_error(y_test , rf_preds)

print(f"Accuracy of training{rf_r2_score_testing}")
print(f"Accuracy of testing{rf_r2_score_traing}")
print(f"MAE{rf_mae}")
```

```python
from sklearn.model_selection import RandomizedSearchCV
 
param_grid = {
'random forest__n_estimators': [100, 200, 300],
'random forest__max_depth': [5, 10, 15],
'random forest__min_samples_split': [2, 5, 10],
'random forest__min_samples_leaf': [1, 2, 4]
}
 
random_search = RandomizedSearchCV(
estimator=rf_pipeline,
param_distributions=param_grid,
n_iter=15,
scoring='r2',
cv=5,
random_state=42,
n_jobs=-1
)
 
random_search.fit(X_train, y_train)
 
print("Best Parameters:", random_search.best_params_)
print("Best Score:", random_search.best_score_)
```

```python
random_search.best_params_
{
    'random forest__n_estimators': 200,
    'random forest__min_samples_split': 5,
    'random forest__min_samples_leaf': 2,
    'random forest__max_depth': 15
    
}
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_preds = rf_best_pipeline.predict(X_test)

print(f"r2 score for random forest after hyperparameter tuning {r2_score(y_test ,rf_best_preds)}")
print(f"MAE score for random foreest after hyperparametr tunung{mean_absolute_error(y_test , rf_best_preds )}")
```

```python
# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_preds))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_preds))

# Calculate RMSE for tuned Random Forest
rf_tuned_preds = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test, rf_tuned_preds))

# Training predictions for tuned Random Forest
rf_tuned_training_preds = random_search.best_estimator_.predict(X_train)
r2_tuned_training = r2_score(y_train, rf_tuned_training_preds)

# MAE for tuned Random Forest
rf_tuned_mae = mean_absolute_error(y_test, rf_tuned_preds)

# R² for tuned Random Forest
rf_tuned_r2 = r2_score(y_test, rf_tuned_preds)

# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest (default)', 'Random Forest (tuned)'],
    'R2': [lr_r2_score_testing, rf_r2_score_testing, rf_tuned_r2],
    'Training R2': [lr_r2_score_traing, rf_r2_score_traing, r2_tuned_training],
    'MAE': [lr_mae, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})
print(results)
```
